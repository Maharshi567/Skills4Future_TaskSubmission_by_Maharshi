```python
#Dataframe

import pandas as pd
import numpy as np

data = {
    "Energy Source": ["Solar", "Wind", "Hydropower", "Geothermal", "Biomass", "Nuclear"],
    "Energy Consumption (MWh)": [1200, np.nan, 2900, np.nan, 2500, 3200],
    "Cost (Million $)": [200, 400, np.nan, 150, 250, np.nan]
}

energy_df = pd.DataFrame(data)
energy_df1 = energy_df.copy()
print(energy_df)
```

```python
energy_df.isnull().sum()

# Remove rows with any missing values
cleaned_df = energy_df.dropna()
subset_df = energy_df.dropna(subset = "Energy Consumption (MWh)")
print("\nData After Removing Rows with Missing Values:")
print(cleaned_df)
print(subset_df)
print(energy_df)
```

```python
# Impute missing values in 'Energy Consumption (MWh)' with the mean
energy_df.iloc[:, 1] = energy_df["Energy Consumption (MWh)"].fillna(energy_df["Energy Consumption (MWh)"].mean())

# Impute missing values in 'Cost (Million $)' with the mean
energy_df.iloc[:, 2] = energy_df["Cost (Million $)"].fillna(energy_df["Cost (Million $)"].mean())

print("\nData After Imputing Missing Values with Mean:")
print(energy_df)
```
